```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler

# Charger le modèle entraîné
model_path = "C:\\Users\\21629\\Desktop\\PFE\\modéle (CNN) 1D basée aal mitbih\\baseline_cnn_mitbih.h5"
model = tf.keras.models.load_model(model_path)

# Charger le signal ECG à tester
signal_path = "C:\\Users\\21629\\Desktop\\PFE\\extraire signal à partir du dataset\\signal_unique2.csv"
signal_df = pd.read_csv(signal_path, header=None)  
signal = signal_df.values.flatten()  # Transformer en tableau 1D

# Vérifier la taille du signal
if signal.shape[0] > 187:
    signal = signal[:187]  # Garder seulement 187 valeurs
elif signal.shape[0] < 187:
    raise ValueError(f"Le signal de test a {signal.shape[0]} valeurs, il en faut 187.")

# Charger les données d'entraînement pour ajuster le MinMaxScaler
train_data_path = "C:\\Users\\21629\\Desktop\\PFE\\modéle (CNN) 1D basée aal mitbih\\mitbih_train.csv\\mitbih_train.csv"
train_data = pd.read_csv(train_data_path)

# Vérifier le nombre de colonnes et exclure la colonne des labels
X_train = train_data.iloc[:, :-1].values  # Exclure la dernière colonne (label)
print("Shape des données d'entraînement :", X_train.shape)

# Normalisation avec MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(X_train)  # Ajuster le scaler sur les données d'entraînement
signal_scaled = scaler.transform(signal.reshape(1, -1))  # Appliquer la normalisation

# Reshape pour correspondre au format attendu par LSTM
signal_scaled = np.expand_dims(signal_scaled, axis=-1)  

# Prédiction
prediction = model.predict(signal_scaled)
predicted_class = np.argmax(prediction)  

# Dictionnaire des classes
labels = {
    0: "Normal",
    1: "Atrial Premature",
    2: "Premature Ventricular Contraction",
    3: "Fusion of Ventricular and Normal",
    4: "Fusion of Paced and Normal"
}

# Affichage du résultat
print(f"Classe prédite : {predicted_class} ({labels.get(predicted_class, 'Inconnu')})")
```
